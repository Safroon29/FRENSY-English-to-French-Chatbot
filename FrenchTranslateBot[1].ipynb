# Import Libraries

This cell imports all the required Python modules and functions used for translation, typing animation, audio playback, and chatbot functionality.

```python
from transformers import MarianMTModel, MarianTokenizer
from gtts import gTTS
from IPython.display import Audio, display
import time
import sys
import os
import re
```

# Load the Translation Model

This cell loads the pre-trained Helsinki-NLP/opus-mt-en-fr model from Hugging Face. The model is specifically designed to translate English text into French.

```python
MODEL_NAME = "Helsinki-NLP/opus-mt-en-fr"

print("⏳ Loading FRENSY's AI translation model...")

tokenizer = MarianTokenizer.from_pretrained(MODEL_NAME)
model = MarianMTModel.from_pretrained(MODEL_NAME)

print("✅ FRENSY is ready!")
```

# Create the Translation Function

This cell defines the function responsible for converting the user's English input into French using the pre-trained translation model

```python
def translate_to_french(text):
    """
    Converts English text into French using
    the Hugging Face MarianMT model.
    """

    inputs = tokenizer(
        text,
        return_tensors="pt",
        padding=True,
        truncation=True
    )

    translated_tokens = model.generate(**inputs)

    french_text = tokenizer.decode(
        translated_tokens[0],
        skip_special_tokens=True
    )

    return french_text
```

# The Typing Effect

This cell creates a character-by-character typing animation to make FRENSY's responses appear more natural and conversational.

```python
def typing_effect(text, delay=0.035):
    """
    Displays text character by character
    to create a typing effect.
    """

    for character in text:
        sys.stdout.write(character)
        sys.stdout.flush()
        time.sleep(delay)

    print()
```

# Translation History

This cell creates a system for storing and displaying previous English-to-French translations made during the chatbot session.

```python
translation_history = []


def show_history():
    """
    Displays all previous translations.
    """

    if not translation_history:
        print("\n📜 No translation history yet.\n")
        return

    print("\n" + "=" * 55)
    print("📜 FRENSY TRANSLATION HISTORY")
    print("=" * 55)

    for i, item in enumerate(translation_history, start=1):

        print(f"\n{i}. 🇬🇧 {item['english']}")
        print(f"   🇫🇷 {item['french']}")

    print("\n" + "=" * 55)
```

# French Pronunciation

This cell uses Google Text-to-Speech (gTTS) to convert the French translation into spoken audio so that users can listen to the pronunciation.

```python
def pronounce_french(text):
    """
    Converts French text into speech.
    """

    try:
        speech = gTTS(
            text=text,
            lang="fr",
            slow=False
        )

        file_name = "frensy_french.mp3"

        speech.save(file_name)

        display(Audio(file_name, autoplay=True))

    except Exception as e:
        print("⚠️ Pronunciation is currently unavailable.")
        print("Please check your internet connection.")
```

# Vocabulary Assistance

This cell provides basic vocabulary assistance by identifying common English words in the user's input and displaying their French equivalents.

```python
common_vocabulary = {
    "hello": "bonjour",
    "hi": "salut",
    "good": "bon",
    "morning": "matin",
    "night": "nuit",
    "thank": "merci",
    "thanks": "merci",
    "please": "s'il vous plaît",
    "friend": "ami",
    "family": "famille",
    "school": "école",
    "college": "université",
    "book": "livre",
    "house": "maison",
    "food": "nourriture",
    "water": "eau",
    "love": "amour",
    "happy": "heureux",
    "beautiful": "beau",
    "goodbye": "au revoir",
    "yes": "oui",
    "no": "non",
    "today": "aujourd'hui",
    "tomorrow": "demain",
    "friend": "ami"
}


def show_vocabulary(text):
    """
    Finds familiar English words in the user's sentence
    and shows their French equivalents.
    """

    words = re.findall(r"[a-zA-Z']+", text.lower())

    found = []

    for word in words:
        if word in common_vocabulary:
            found.append(
                (word, common_vocabulary[word])
            )

    if found:

        print("\n📚 Vocabulary Help")
        print("-" * 35)

        for english, french in found:
            print(f"🇬🇧 {english}  →  🇫🇷 {french}")

        print("-" * 35)
```

```python
def show_banner():

    print("\n")
    print("╔" + "═" * 58 + "╗")
    print("║" + " " * 18 + "🇬🇧 FRENSY 🇫🇷" + " " * 19 + "          ║")
    print("║" + " " * 8 + "Your AI English-to-French Buddy" + " " * 19 + "║")
    print("╚" + "═" * 58 + "╝")

    print("\n✨ Translate • Learn • Connect ✨")
```

# Greeting Responses

This cell allows FRENSY to recognize common greetings such as "hello", "hi", and "good morning" and respond with appropriate French greetings.

```python
def handle_greeting(text):

    text_lower = text.lower().strip()

    greetings = {
        "hi": "Bonjour ! 👋",
        "hello": "Bonjour ! 👋",
        "hey": "Salut ! 😄",
        "good morning": "Bonjour ! ☀️",
        "good evening": "Bonsoir ! 🌆",
        "good night": "Bonne nuit ! 🌙"
    }

    return greetings.get(text_lower)
```

# Main Chatbot

This cell combines the translation, typing animation, history, pronunciation, vocabulary, and command features into the main FRENSY chatbot.

```python
def start_frensy():

    show_banner()

    print("\n🤖 Welcome! I'm FRENSY.")
    print("I can translate English into French for you.")

    print("\nCommands:")
    print("  📜 history  → View previous translations")
    print("  🔊 speak    → Hear the last French translation")
    print("  ❌ exit     → Close FRENSY")

    last_translation = None

    while True:

        print("\n" + "-" * 60)

        user_input = input("👤 You: ").strip()

        if not user_input:
            print("🤖 FRENSY: Please enter something for me to translate!")
            continue

        # EXIT
        if user_input.lower() == "exit":

            print("\n🤖 FRENSY: Au revoir! 👋")
            print("Thanks for using FRENSY! 🇬🇧 → 🇫🇷")
            break

        # HISTORY
        if user_input.lower() == "history":

            show_history()
            continue

        # SPEAK
        if user_input.lower() == "speak":

            if last_translation:

                print("🔊 Playing French pronunciation...")
                pronounce_french(last_translation)

            else:

                print("🤖 FRENSY: There is no translation to speak yet.")

            continue

        # GREETING
        greeting = handle_greeting(user_input)

        if greeting:

            print("\n🤖 FRENSY: ", end="")
            typing_effect(greeting)

            last_translation = greeting

            continue

        # TRANSLATION
        print("\n🤖 FRENSY is typing... ⌨️")

        french_translation = translate_to_french(user_input)

        last_translation = french_translation

        # Save history
        translation_history.append({
            "english": user_input,
            "french": french_translation
        })

        print("🇫🇷 FRENSY: ", end="")
        typing_effect(french_translation)

        # Vocabulary
        show_vocabulary(user_input)
```

# FRENSY

This final cell starts the FRENSY chatbot and allows the user to enter English sentences and receive their French translations interactively.

```python
start_frensy()
```
